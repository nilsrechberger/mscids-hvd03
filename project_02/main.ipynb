## Data Prparation

### Load data

```python
import pandas as pd

df = pd.read_csv(
    filepath_or_buffer='./data/project_02_version_19_dataset.csv',
    dtype={
        'Page Viewed': int,
        'Product Views': int,
        'Session Duration (seconds)': int,
        'Cart Value (CHF)': float,
        'Shipping Cost Shown (CHF)': float,
        'Order Value (CHF)': float
    },
    parse_dates=[
        'Session Start'
    ]
)

# Preview
df.head()
```

### Stats

```python
df.describe(include='all')
```

## Questions

### Question 01

```python
df_funnel_chart = df[
    [
        'Session ID',
        'Added to Cart',
        'Checkout Started',
        'Purchased'
    ]
]

print(f'Total Sessions = {df_funnel_chart['Session ID'].count()}')
df_funnel_chart[
    [   'Added to Cart',
        'Checkout Started',
        'Purchased'
    ]
].value_counts(dropna=True)
```

### Question 02

```python
df_stacked_bars = df[
    [
        'Session ID',
        'Session Start',
        'Traffic Source'
    ]
]

df_stacked_bars['Session Start'] = df_stacked_bars['Session Start'].dt.month

stacked_data = (
    df_stacked_bars
    .groupby(['Session Start', 'Traffic Source'])['Session ID']
    .count()
    .unstack(fill_value=0)
)

stacked_data
```

### Question 03

```python
df_hist = df[
    [
        'Session Duration (seconds)'
    ]
]

df_hist.plot.hist(bins=50)
```

### Question 04

```python
df_log_reg = df[
    [
        'Session Duration (seconds)',
        'Purchased'
    ]
]

df_log_reg.plot.scatter(
    x='Session Duration (seconds)',
    y='Purchased'
)
```

### Question 05

```python
df_grouped_bars = df[
    [
        'Country',
        'Traffic Source',
        'Purchased'
    ]
]

grouped_data = (
    df_grouped_bars
    .groupby(['Country', 'Traffic Source'])['Country']
    .count()
    .unstack(fill_value=0)
)

grouped_data
```
