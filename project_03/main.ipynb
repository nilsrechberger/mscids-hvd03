# U.S. Data Centers

> Source: https://makeovermonday.vercel.app/dataset/us-data-center-locations

## Read data

```python
import pandas as pd

df = pd.read_csv('./data/data_center_loc.csv')

# preview
df.head()
```

## Stats

```python
df.describe()
```

```python
df.value_counts(
    [
        'city',
        'county',
        'status',
        'location_confidence'
    ]
)
```

## Visualizations

### Top 5 Cities

```python
cities = df['city'].value_counts()
cities.head().plot.bar()
```

### Status Ratio

```python
status = df['status'].value_counts()

status.plot.pie()
```

### Locations

```python
df.plot.scatter(
    x='long',
    y='lat',
    c='status'
)
```
